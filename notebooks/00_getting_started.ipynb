```python
# Course setup - run once when this Colab runtime starts.
# Installs the pinned Python environment. Does nothing when run locally.
import subprocess, sys

try:
    import google.colab  # noqa: F401
    IN_COLAB = True
except ModuleNotFoundError:
    IN_COLAB = False

if IN_COLAB:
    subprocess.run(
        ["wget", "-q",
         "https://raw.githubusercontent.com/kaancet/2026KU-invivo/main/setup_colab.py"],
        check=True,
    )
    subprocess.run([sys.executable, "setup_colab.py"], check=True)
```

```python
# Course data - mounted from Google Drive (data only; ~5-10 GB).
# One-time per Google account: open the shared data-folder link and
# choose "Organise > Add shortcut to Drive".
import sys
from pathlib import Path

try:
    import google.colab  # noqa: F401
    from google.colab import drive
    drive.mount("/content/drive")
    DATA_ROOT = Path("/content/drive/MyDrive/2026KU-invivo-data")
except ModuleNotFoundError:
    DATA_ROOT = Path(sys.executable).parents[2] / "data"

DATA_PATH = DATA_ROOT / "processed"
assert DATA_PATH.exists(), (
    f"Data not found: {DATA_PATH} "
    "(add the shared Drive folder shortcut / check the folder name)"
)
print("Data path:", DATA_PATH)
```
